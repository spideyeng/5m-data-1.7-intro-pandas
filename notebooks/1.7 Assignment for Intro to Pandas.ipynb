# Question 5
*How do you create a DataFrame from a nested dictionary of dictionaries?*

## Answer 5

```python
# Create a nested dictionary of dictionaries
nested_dict = {
    'student1': {
        'name': 'John',
        'age': 20,
        'grades': {'math': 90, 'science': 95}
    },
    'student2': {
        'name': 'Emma',
        'age': 21,
        'grades': {'math': 88, 'science': 92}
    }
}

# Create DataFrame from nested dictionary
df = pd.DataFrame.from_dict(nested_dict, orient='index')
df
```

```python
# To access the nested 'grades' dictionary, we can use json_normalize
from pandas import json_normalize

# Normalize the grades column
grades_df = pd.json_normalize(df['grades'])
grades_df.index = df.index  # Keep the same index as the main DataFrame

# Now combine with the other columns
result = pd.concat([df[['name', 'age']], grades_df], axis=1)
result
```
